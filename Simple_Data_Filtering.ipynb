```python
students = [
    {"name": "Asha", "age": 20, "marks": 85, "department": "CSE"},
    {"name": "Rahul", "age": 21, "marks": 65, "department": "ECE"},
    {"name": "Priya", "age": 19, "marks": 92, "department": "CSE"},
    {"name": "Arun", "age": 22, "marks": 45, "department": "MECH"},
    {"name": "Sneha", "age": 20, "marks": 78, "department": "ECE"}
]

print(students)
```

```python
high_marks = []

for student in students:
    if student["marks"] > 75:
        high_marks.append(student)

print("Students who scored more than 75 marks:")
print(high_marks)
```

```python
cse_students = []

for student in students:
    if student["department"] == "CSE":
        cse_students.append(student)

print("Students in the CSE department:")
print(cse_students)
```

```python
young_students = []

for student in students:
    if student["age"] < 21:
        young_students.append(student)

print("Students younger than 21:")
print(young_students)
```

```python
print("Total number of students:", len(students))
print("Students scoring above 75:", len(high_marks))
print("Number of CSE students:", len(cse_students))
print("Students younger than 21:", len(young_students))
```
